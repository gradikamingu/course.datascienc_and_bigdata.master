```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Age": [25,30,22,28,35,40,29,25,np.nan,32],
    "Salaire": [2000,2500,np.nan,2700,3000,3500,40000,2000,2800,3100],
    "Score": [80,75,60,90,np.nan,85,70,80,95,65],
    "Sexe": ["Homme","Femme","Homme","Femme","Homme","Homme","Femme","Homme","Femme","Homme"],
    "Ville": ["Kinshasa","Lubumbashi","Goma","Kinshasa","Kisangani","Lubumbashi","Goma","Kinshasa","Lubumbashi","Kinshasa"],
    "Departement": ["IT","Finance","RH",np.nan,"IT","Finance","RH","IT","Finance","IT"]
}
```

```python
data_employee = pd.DataFrame(data)
```

```python
data_employee
```

# Gestion des données manquantes

## 1. Liste des valeurs manquantes par variable

```python
data_employee.isnull().sum()
```

## 2. Liste des types des variables

```python
data_quantitatives = data_employee.select_dtypes(include=[np.number]).columns.tolist()
data_qualitatives = data_employee.select_dtypes(exclude=[np.number]).columns.tolist()
```

```python
data_quantitatives
```

```python
data_qualitatives
```

```python
data_employee["Age"].fillna(data_employee["Age"].mean(), inplace = True )
```

```python
data_employee["Salaire"].fillna(data_employee["Salaire"].mean(), inplace = True )
```

```python
data_employee["Score"].fillna(data_employee["Score"].mean(), inplace = True )
```

```python
data_employee["Departement"].fillna(data_employee["Departement"].mode()[0], inplace = True )
```
